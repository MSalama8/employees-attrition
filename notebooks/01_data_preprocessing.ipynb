# Employee Attrition — Data Preprocessing


```python
import pandas as pd
import numpy as np
```

## Read the CSV


```python
df = pd.read_csv('../data/raw/employee_attrition_raw.csv')
df.head()
```

## 1. Understand the data


```python
print('Rows and columns:', df.shape)
df.info()
```

## 2. Summary statistics


```python
df.describe(include='number')
```

```python
df.describe(include='object')
```

## 3. Check missing values and duplicates


```python
df.isna().sum()
```

```python
print('Duplicated rows:', df.duplicated().sum())
print('Duplicated employee numbers:', df['EmployeeNumber'].duplicated().sum())
```

## 4. Check Attrition


```python
df.groupby('Attrition')['EmployeeNumber'].count()
```

## 5. Cell for Checking Values Before Deletion

```python
print("EmployeeCount:", np.unique(df["EmployeeCount"]))
print("Over18:", np.unique(df["Over18"]))
print("StandardHours:", np.unique(df["StandardHours"]))

unique_employee_numbers = np.unique(df["EmployeeNumber"])

print("Total rows:", len(df))
print("Unique Employee Numbers:", len(unique_employee_numbers))
print("Duplicate Employee Numbers:", len(df) - len(unique_employee_numbers))
```

## 6. Remove constant columns


```python
df.drop(columns=['EmployeeCount', 'Over18', 'StandardHours'], inplace=True)
print('Shape after cleaning:', df.shape)
df.head()
```

## 7. Check the result

```python
df.isna().sum()
```

```python
print('Duplicated rows after cleaning:', df.duplicated().sum())
print('Duplicated employee numbers after cleaning:', df['EmployeeNumber'].duplicated().sum())
```

## 8. Save the cleaned CSV


```python
df.to_csv('../data/processed/employee_attrition_cleaned.csv', index=False)
print('Cleaned CSV saved.')
```
